```python
df['image_exists'] = df['image_id'].apply(
    lambda x: os.path.exists(
        f"C:/Users/jagad/Documents/my_classes/Tasks/projects_of_guvi/Final_project_Deep_learning/data/raw_images/{x}.jpg"
    )
)

cleaned_df = df[df['image_exists'] == True].copy()

print("Original rows:", len(df))
print("After removing failed downloads:", len(cleaned_df))

cleaned_df.to_csv(r"C:\Users\jagad\Documents\my_classes\Tasks\projects_of_guvi\Final_project_Deep_learning\data\cleaned_dataset.csv", index=False)

print("Saved as data/cleaned_dataset.csv")
```

## Image Verification

```python
from PIL import Image

# Read the clean dataset
df = pd.read_csv(r"C:\Users\jagad\Documents\my_classes\Tasks\projects_of_guvi\Final_project_Deep_learning\data\cleaned_dataset.csv")

corrupted_ids = []

for image_id in df['image_id']:
    file_path = f"C:/Users/jagad/Documents/my_classes/Tasks/projects_of_guvi/Final_project_Deep_learning/data/raw_images/{image_id}.jpg"
    try:
        img = Image.open(file_path)
        img.verify()   # checks if the image file is valid, without fully loading it
    except Exception as e:
        corrupted_ids.append(image_id)


print("Total images checked:", len(df))
print("Corrupted images found:", len(corrupted_ids))


# Remove corrupted rows and save final clean dataset
final_df = df[~df['image_id'].isin(corrupted_ids)]
final_df.to_csv(r"C:\Users\jagad\Documents\my_classes\Tasks\projects_of_guvi\Final_project_Deep_learning\data\final_dataset.csv", index=False)

print("Saved as data/final_dataset.csv with", len(final_df), "rows")
```
